```python
# plot the results of Riemannian meta learning
import json

# Open and read the JSON file
with open('res_Riemannian.json', 'r') as file:
    res_r = json.load(file)
    
with open('res_Euclidean.json', 'r') as file:
    res_e = json.load(file)
    
metrics = list(res_r.keys())
metrics
```

```python
import matplotlib.pyplot as plt

for i in range(1, len(res_r.keys())):
    fig = plt.figure()
    ax = fig.gca()
    
    ax.plot(res_e['total_time'], res_e[metrics[i]], label='Projected SBO')
    ax.plot(res_r['total_time'], res_r[metrics[i]], label='RieSBO')

    ax.set_xlabel("CPU time")
    ax.set_ylabel(metrics[i])
    ax.legend()
    fig.show()
    fig.savefig('plots/meta_learning_' + metrics[i] + '.pdf')
```
